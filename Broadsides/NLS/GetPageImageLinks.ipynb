```python
from lxml import etree

# Load all mets xsmls files and parse them

volumes_filenames = open('broadsides.txt', 'r').read().splitlines()
volume_ids = {}
for vol_filename in volumes_filenames:
    volume_id = vol_filename.split('/')[-1]
    mets_filename = f"{vol_filename}/{volume_id}-mets.xml"
    print(mets_filename)
    mets_tree = etree.parse(mets_filename)
    volume_ids[volume_id] = mets_tree
```

```python
namespaces = {
            "mods": 'http://www.loc.gov/mods/v3',
            "mets": 'http://www.loc.gov/METS/',
            "xlink": 'http://www.w3.org/1999/xlink'
        }
```

```python
def get_page_info(mets_tree):
    # Find all page number and file id
    pages = []
    for div in mets_tree.xpath("//mets:div[@ORDER]", namespaces=namespaces):
        page_number = div.attrib['ORDER']
        # Look for <fptr> elements inside this <div> with FILEID ending in '.30'
        for fptr in div:
            file_id = fptr.attrib['FILEID']
            if file_id[-3:] == ".30":
                #print(f"page number: {page_number}, id: {image_file_id}")
                pages.append({
                    "number": page_number,
                    "image_file_id": file_id
                })
                break
    #print(pages)
    # Find the image url
    for page in pages:
        page_id = page['image_file_id']
        del page['image_file_id']
        page_image_file = mets_tree.xpath(f"//mets:file[@ID = '{page_id}']", namespaces=namespaces)[0]
        #print(page_image_file[0])
        page_image_url = page_image_file[0].get('{%s}href' % namespaces['xlink'])
        page['image_url'] = page_image_url
    return pages
```

```python
get_page_info(volume_ids['103831246'])
```

```python
def get_mmsid(mets_tree):
    result = mets_tree.xpath("//mods:recordIdentifier/text()", namespaces=namespaces)
    if not result:
        return None
    return result[0]
```

```python
get_mmsid(volume_ids['103831246'])
```

```python
volume_page_urls = {}
for volume_id in volume_ids:
    mets_tree = volume_ids[volume_id]
    volume_page_urls[volume_id] = {}
    pages = get_page_info(mets_tree)
    print(f"volume:{volume_id}")
    mmsid = get_mmsid(mets_tree)
    print(f"mmsid:{mmsid}")
    volume_page_urls[volume_id]["mmsid"] = mmsid
    volume_page_urls[volume_id]["pages"] = pages

volume_page_urls
```

```python
import json

with open('volume_page_image_urls.json', "w") as f:
    json.dump(volume_page_urls, f)
```
